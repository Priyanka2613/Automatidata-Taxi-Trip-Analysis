### Boxplots

Perform a check for outliers on relevant columns such as trip distance and trip duration. Remember, some of the best ways to identify the presence of outliers in data are box plots and histograms. 

**Note:** Remember to convert your date columns to datetime in order to derive total trip duration.  

```python
df['tpep_pickup_datetime'] = pd.to_datetime(df['tpep_pickup_datetime'])
df['tpep_dropoff_datetime'] = pd.to_datetime(df['tpep_dropoff_datetime'])

# Calculate trip duration in minutes
df['trip_duration'] = (
    df['tpep_dropoff_datetime'] - df['tpep_pickup_datetime']
).dt.total_seconds() / 60

df[['trip_duration', 'trip_distance']].head()
```

**trip distance**

```python
plt.figure(figsize=(10, 4))
sns.boxplot(x=df['trip_distance'])
plt.title('Distribution of Trip Distance')
plt.xlabel('Trip Distance (miles)')
plt.show()
```

```python
# Create histogram of trip_distance
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='trip_distance', bins=40)
plt.title('Distribution of Trip Distance')
plt.xlabel('Trip Distance (miles)')
plt.ylabel('Number of Trips')
plt.show()
```

**total amount**

```python
# Create box plot of total_amount
plt.figure(figsize=(10, 4))
sns.boxplot(x=df['total_amount'])
plt.title('Distribution of Total Amount')
plt.xlabel('Total Amount ($)')
plt.show()
```

```python
# Create histogram of total_amount
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='total_amount', bins=40)
plt.title('Distribution of Total Amount')
plt.xlabel('Total Amount ($)')
plt.ylabel('Number of Trips')
plt.show()
```

**tip amount**

```python
# Create box plot of tip_amount
plt.figure(figsize=(10, 4))
sns.boxplot(x=df['tip_amount'])
plt.title('Distribution of Tip Amount')
plt.xlabel('Tip Amount ($)')
plt.show()
```

```python
# Create histogram of tip_amount
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='tip_amount', bins=40)
plt.title('Distribution of Tip Amount')
plt.xlabel('Tip Amount ($)')
plt.ylabel('Number of Trips')
plt.show()
```

**tip_amount by vendor**

```python
# Create histogram of tip_amount by vendor
plt.figure(figsize=(10, 5))
sns.histplot(
    data=df,
    x='tip_amount',
    hue='VendorID',
    bins=40,
    multiple='layer'
)
plt.title('Tip Amount Distribution by Vendor')
plt.xlabel('Tip Amount ($)')
plt.ylabel('Number of Trips')
plt.show()
```

**Mean tips by passenger count**

Examine the unique values in the `passenger_count` column.

```python
# Examine unique passenger counts
df['passenger_count'].value_counts().sort_index()
```

```python
# Calculate mean tips by passenger count
mean_tips = (
    df.groupby('passenger_count')['tip_amount']
    .mean()
    .reset_index()
)

mean_tips
```

**Plot total ride count by month**

Begin by calculating total ride count by month.

```python
# Get total number of rides for each month
monthly_rides = df['month'].value_counts()

monthly_rides
```

To confirm your conclusion, consider the following experiment:
1. Create a sample of coordinates from a normal distribution&mdash;in this case 1,500 pairs of points from a normal distribution with a mean of 10 and a standard deviation of 5
2. Calculate the distance between each pair of coordinates 
3. Group the coordinates by endpoint and calculate the mean distance between that endpoint and all other points it was paired with
4. Plot the mean distance for each unique endpoint

```python
#BONUS CONTENT

#1. Generate random points on a 2D plane from a normal distribution
#==> ENTER YOUR CODE HERE

# 1. Generate random points on a 2D plane
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate 1,500 pairs of 2D points
points = np.random.normal(loc=10, scale=5, size=(3000, 2))

# Check the shape and first five points
print("Shape of points:", points.shape)
print(points[:5])

# 2. Calculate Euclidean distances between points in first half and second half of array
#==> ENTER YOUR CODE HERE

# Split points into two groups
first_half = points[:1500]
second_half = points[1500:]

# Calculate Euclidean distances
distances = np.linalg.norm(first_half - second_half, axis=1)

# Display results
print("Number of distances:", len(distances))
print("First five distances:", distances[:5])

# 3. Group the coordinates by "drop-off location", compute mean distance
#==> ENTER YOUR CODE HERE

# Create a DataFrame with endpoint coordinates and distances
simulation_df = pd.DataFrame({
    'dropoff_x': second_half[:, 0],
    'dropoff_y': second_half[:, 1],
    'distance': distances
})

# Group by endpoint coordinates and calculate mean distance
mean_distances = simulation_df.groupby(
    ['dropoff_x', 'dropoff_y']
)['distance'].mean()

# Display results
print("Number of unique endpoints:", len(mean_distances))
print(mean_distances.head())

# 4. Plot the mean distance between each endpoint ("drop-off location") and all points it connected to
#==> ENTER YOUR CODE HERE

# Sort mean distances in ascending order
sorted_distances = mean_distances.sort_values()

# Create a bar plot
plt.figure(figsize=(14, 6))

plt.bar(
    range(len(sorted_distances)),
    sorted_distances.values,
    color='steelblue'
)

plt.title('Mean Distance by Simulated Drop-off Location')
plt.xlabel('Drop-off Locations (Sorted by Mean Distance)')
plt.ylabel('Mean Euclidean Distance')
plt.xticks([])
plt.tight_layout()
plt.show()
```

**Histogram of rides by drop-off location**

First, check to whether the drop-off locations IDs are consecutively numbered. For instance, does it go 1, 2, 3, 4..., or are some numbers missing (e.g., 1, 3, 4...). If numbers aren't all consecutive, the histogram will look like some locations have very few or no rides when in reality there's no bar because there's no location. 

```python
# Check if all drop-off locations are consecutively numbered

# Get sorted unique drop-off location IDs
location_ids = sorted(df['DOLocationID'].dropna().unique())

# Check for missing IDs between the minimum and maximum
all_ids = set(range(int(min(location_ids)), int(max(location_ids)) + 1))
missing_ids = sorted(all_ids - set(location_ids))

print("Number of unique drop-off locations:", len(location_ids))
print("Minimum location ID:", min(location_ids))
print("Maximum location ID:", max(location_ids))
print("Missing location IDs:", missing_ids)
print("Are IDs consecutive?", len(missing_ids) == 0)
```

To eliminate the spaces in the historgram that these missing numbers would create, sort the unique drop-off location values, then convert them to strings. This will make the histplot function display all bars directly next to each other. 

```python

# Histogram of rides by drop-off location

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Sort unique drop-off location IDs in ascending order
sorted_ids = sorted(df['DOLocationID'].dropna().unique())

# Convert location IDs to strings
location_labels = [str(int(i)) for i in sorted_ids]

# Convert drop-off location IDs to strings for plotting
dropoff_locations = df['DOLocationID'].dropna().astype(int).astype(str)

# Create histogram of rides by drop-off location
plt.figure(figsize=(16, 6))

sns.countplot(
    x=dropoff_locations,
    order=location_labels,
    color='steelblue'
)

# Add chart title and labels
plt.title('Number of Taxi Rides by Drop-off Location')
plt.xlabel('Drop-off Location ID')
plt.ylabel('Number of Rides')

# Hide crowded location labels
plt.xticks([])

# Adjust layout and display
plt.tight_layout()
plt.show()
```
